```python
import numpy as np
import pandas as pd

pd.set_option('display.max_columns', 60)

raw_path = '../data/raw/automobileEDA_dirty_training.csv'
df_raw = pd.read_csv(raw_path)

print(f"Dimensi data mentah: {df_raw.shape[0]} baris x {df_raw.shape[1]} kolom")
df_raw.head(3)
```

```python
# 1. Lima baris pertama dataset
print("=== 5 BARIS PERTAMA DATASET ===")
display(df_raw.head())

# 2. Jumlah baris dan kolom
n_rows, n_cols = df_raw.shape
print(f"\n=== DIMENSI DATASET ===")
print(f"Jumlah Baris : {n_rows}")
print(f"Jumlah Kolom : {n_cols}")

# 3. Nama setiap kolom & 4. Tipe data setiap kolom
print("\n=== SKEMA & TIPE DATA SETIAP KOLOM ===")
dtype_df = pd.DataFrame({
    'Nama Kolom': df_raw.columns,
    'Tipe Data': df_raw.dtypes.values
})
display(dtype_df)

# 5. Jumlah missing values pada setiap kolom
print("\n=== MISSING VALUES PER KOLOM ===")
missing_df = pd.DataFrame({
    'Missing (Count)': df_raw.isna().sum(),
    'Missing (%)': (df_raw.isna().sum() / n_rows * 100).round(2)
})
missing_only = missing_df[missing_df['Missing (Count)'] > 0]
display(missing_only)

# 6. Jumlah duplicate records
n_dups = df_raw.duplicated().sum()
print(f"\n=== DUPLICATE RECORDS ===")
print(f"Jumlah Baris Duplikat Penuh: {n_dups}")

# 7. Nilai unik pada kolom kategorikal relevan
print("\n=== NILAI UNIK KOLOM KATEGORIKAL ===")
cat_cols = ['make', 'aspiration', 'num-of-doors', 'body-style', 
            'drive-wheels', 'engine-location', 'engine-type', 
            'num-of-cylinders', 'fuel-system', 'horsepower-binned']

for col in cat_cols:
    uniques = df_raw[col].dropna().unique().tolist()
    print(f"{col} ({len(uniques)} unik):")
    print(f"  {uniques}\n")
```

```python
# ==========================================
# METRIK SEBELUM PEMBERSIHAN (BEFORE)
# ==========================================
rows_before = len(df_raw)
cols_before = len(df_raw.columns)
missing_before = int(df_raw.isna().sum().sum())
dups_before = int(df_raw.duplicated().sum())

print(f"{'='*20} METRIK SEBELUM CLEANING {'='*20}")
print(f"Dimensi Data       : {rows_before} baris x {cols_before} kolom")
print(f"Total Missing Value: {missing_before}")
print(f"Baris Duplikat     : {dups_before}")
print("=" * 64)

# ==========================================
# PROSES PEMBERSIHAN DATA (CLEANING)
# ==========================================
df_clean = df_raw.copy()
modified_columns = []

# 1. Hapus kolom tidak relevan (transaction_date)
df_clean = df_clean.drop(columns=['transaction_date'])
modified_columns.append('transaction_date (dihapus)')

# 2. Hapus baris duplikat penuh
df_clean = df_clean.drop_duplicates().reset_index(drop=True)

# 3. Standarisasi teks: trim spasi liar & lowercase
str_cols = df_clean.select_dtypes(include=['object', 'string']).columns.tolist()
for col in str_cols:
    df_clean[col] = df_clean[col].astype(str).str.strip().str.lower()
    df_clean[col] = df_clean[col].replace(['nan', 'none', ''], np.nan)
    modified_columns.append(f"{col} (strip & lowercase)")

# 4. Imputasi Missing Values
# Numerik: Median (tahan outlier)
numeric_impute = ['price', 'horsepower', 'stroke']
for col in numeric_impute:
    med_val = df_clean[col].median()
    df_clean[col] = df_clean[col].fillna(med_val)
    modified_columns.append(f"{col} (imputasi median={med_val})")

# Kategorikal: Modus (kelas mayoritas)
doors_mode = df_clean['num-of-doors'].mode()[0]
df_clean['num-of-doors'] = df_clean['num-of-doors'].fillna(doors_mode)
modified_columns.append(f"num-of-doors (imputasi modus='{doors_mode}')")

make_mode = df_clean['make'].mode()[0]
df_clean['make'] = df_clean['make'].fillna(make_mode)
modified_columns.append(f"make (imputasi modus='{make_mode}')")

df_clean['horsepower-binned'] = df_clean['horsepower-binned'].fillna('low')
modified_columns.append("horsepower-binned (imputasi='low')")

# ==========================================
# METRIK SESUDAH PEMBERSIHAN (AFTER)
# ==========================================
rows_after = len(df_clean)
cols_after = len(df_clean.columns)
missing_after = int(df_clean.isna().sum().sum())
dups_after = int(df_clean.duplicated().sum())

print(f"\n{'='*20} METRIK SESUDAH CLEANING {'='*20}")
print(f"Dimensi Data       : {rows_after} baris x {cols_after} kolom")
print(f"Total Missing Value: {missing_after}")
print(f"Baris Duplikat     : {dups_after}")
print(f"Duplikat Dihapus   : {dups_before - dups_after} baris")
print("=" * 64)

# Tabel Ringkasan Komparasi Sebelum vs Sesudah
comparison_df = pd.DataFrame({
    'Metrik': ['Jumlah Baris', 'Jumlah Kolom', 'Total Missing Values', 'Baris Duplikat'],
    'Sebelum (Raw)': [rows_before, cols_before, missing_before, dups_before],
    'Sesudah (Clean)': [rows_after, cols_after, missing_after, dups_after],
    'Perubahan': [rows_after - rows_before, cols_after - cols_before, missing_after - missing_before, -dups_before]
})
print("\n=== RINGKASAN PERBANDINGAN SEBELUM VS SESUDAH ===")
display(comparison_df)

print("\n=== DAFTAR KOLOM YANG MENGALAMI PERUBAHAN ===")
for item in set(modified_columns):
    print(f"- {item}")
```

```python
# ==========================================
# SIMPAN CONTOH NILAI SEBELUM TRANSFORMASI
# ==========================================
before_examples = {
    'num-of-doors': df_clean['num-of-doors'].head(5).tolist(),
    'num-of-cylinders': df_clean['num-of-cylinders'].head(5).tolist(),
    'horsepower-binned': df_clean['horsepower-binned'].head(5).tolist(),
    'make': df_clean['make'].head(5).tolist(),
    'price': df_clean['price'].head(5).tolist(),
    'horsepower': df_clean['horsepower'].head(5).tolist(),
    'body-style': df_clean['body-style'].head(5).tolist(),
}

df_transformed = df_clean.copy()

# ==========================================
# 1. BINARY & ORDINAL ENCODING
# ==========================================
# num-of-doors: two -> 0.0, four -> 1.0
door_map = {'two': 0.0, 'four': 1.0}
df_transformed['num-of-doors'] = df_transformed['num-of-doors'].map(door_map).astype(float)

# num-of-cylinders: konversi kata ke numerik lalu Min-Max (cyl - 2) / 10
cyl_map = {
    'two': 2, 'three': 3, 'four': 4, 'five': 5,
    'six': 6, 'eight': 8, 'twelve': 12
}
df_transformed['num-of-cylinders'] = (df_transformed['num-of-cylinders'].map(cyl_map) - 2) / 10.0

# horsepower_ordinal: low -> 0, medium -> 1, high -> 2
hp_bin_map = {'low': 0, 'medium': 1, 'high': 2}
df_transformed['horsepower_ordinal'] = (
    df_transformed['horsepower-binned'].map(hp_bin_map).fillna(0).astype(int)
)

# ==========================================
# 2. FREQUENCY ENCODING (TARGETLESS)
# ==========================================
# make_freq: proporsi kemunculan merk terhadap total populasi data
make_counts = df_transformed['make'].value_counts()
df_transformed['make_freq'] = df_transformed['make'].map(make_counts) / len(df_transformed)

# ==========================================
# 3. MIN-MAX SCALING (NUMERIK CONTINUOUS)
# ==========================================
scale_cols = [
    'symboling', 'curb-weight', 'engine-size', 'horsepower',
    'peak-rpm', 'city-mpg', 'highway-mpg', 'price'
]
for col in scale_cols:
    c_min = df_transformed[col].min()
    c_max = df_transformed[col].max()
    df_transformed[col] = (df_transformed[col] - c_min) / (c_max - c_min)

# ==========================================
# 4. ONE-HOT ENCODING (NOMINAL CATEGORICAL)
# ==========================================
cat_ohe_cols = [
    'body-style', 'drive-wheels', 'aspiration',
    'engine-type', 'engine-location', 'fuel-system'
]
for col in cat_ohe_cols:
    dummies = pd.get_dummies(df_transformed[col], prefix=col, dtype=int)
    df_transformed = pd.concat([df_transformed, dummies], axis=1)

# Drop kolom asal yang telah ditransformasi
df_transformed = df_transformed.drop(columns=cat_ohe_cols + ['make', 'horsepower-binned'])

# ==========================================
# 5. SUSUN URUTAN 50 KOLOM REFERENSI
# ==========================================
target_columns = [
    'symboling', 'normalized-losses', 'num-of-doors', 'wheel-base', 'length',
    'width', 'height', 'curb-weight', 'num-of-cylinders', 'engine-size',
    'bore', 'stroke', 'compression-ratio', 'horsepower', 'peak-rpm',
    'city-mpg', 'highway-mpg', 'price', 'city-L/100km', 'diesel', 'gas',
    'horsepower_ordinal', 'body-style_convertible', 'body-style_hardtop',
    'body-style_hatchback', 'body-style_sedan', 'body-style_wagon',
    'drive-wheels_4wd', 'drive-wheels_awd', 'drive-wheels_fwd', 'drive-wheels_rwd',
    'aspiration_std', 'aspiration_turbo', 'engine-type_dohc', 'engine-type_l',
    'engine-type_ohc', 'engine-type_ohcf', 'engine-type_ohcv', 'engine-type_rotor',
    'engine-location_front', 'engine-location_rear', 'fuel-system_1bbl',
    'fuel-system_2bbl', 'fuel-system_4bbl', 'fuel-system_idi', 'fuel-system_mfi',
    'fuel-system_mpfi', 'fuel-system_spdi', 'fuel-system_spfi', 'make_freq'
]

df_transformed = df_transformed[target_columns]

print(f"Dimensi Setelah Transformasi: {df_transformed.shape[0]} baris x {df_transformed.shape[1]} kolom")

# ==========================================
# TABEL KOMPARASI SEBELUM VS SESUDAH
# ==========================================
comp_table = pd.DataFrame({
    'Fitur': ['num-of-doors', 'num-of-cylinders', 'horsepower-binned', 'make', 'price', 'horsepower'],
    'Sebelum (Clean)': [
        str(before_examples['num-of-doors'][:3]),
        str(before_examples['num-of-cylinders'][:3]),
        str(before_examples['horsepower-binned'][:3]),
        str(before_examples['make'][:3]),
        str([round(x, 1) for x in before_examples['price'][:3]]),
        str([round(x, 1) for x in before_examples['horsepower'][:3]]),
    ],
    'Sesudah (Transformed)': [
        str(df_transformed['num-of-doors'].head(3).tolist()),
        str(df_transformed['num-of-cylinders'].head(3).tolist()),
        str(df_transformed['horsepower_ordinal'].head(3).tolist()) + " (horsepower_ordinal)",
        str([round(x, 4) for x in df_transformed['make_freq'].head(3).tolist()]) + " (make_freq)",
        str([round(x, 4) for x in df_transformed['price'].head(3).tolist()]),
        str([round(x, 4) for x in df_transformed['horsepower'].head(3).tolist()]),
    ]
})

print("\n=== CONTOH SAMPEL SEBELUM VS SESUDAH TRANSFORMASI ===")
display(comp_table)
```
